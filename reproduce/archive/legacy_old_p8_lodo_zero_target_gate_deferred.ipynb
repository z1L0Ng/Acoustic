## Verified Contract / Proposed Gate / HOLD

**Verified Contract**
- 合法 zero-target 必须无 target labels、无 target head fitting、无 target threshold/selection；outer labels 只能在固定预测完成后 terminal scoring。
- dataset-native heads 的 label spaces 不同，不能凭名称直接跨 target 使用。

**Proposed Gate**
- 只允许 source-trained shared crackle/wheeze compatible surface；对 target 直接预测，不创建 target-native head。
- 每个 target 必须先有 frozen compatibility/eligibility/mapping receipt，且 source selection 不使用 target outcome。

**HOLD**
- 当前四数据集没有冻结、覆盖充分且语义合法的统一 shared target surface。HF 仅 observed-positive partial supervision；KAUH shared mapping/diagnosis HOLD。
- 因此 P8 不得执行；任何 target-head fitting 都应路由到 P7 并命名 adaptation。
