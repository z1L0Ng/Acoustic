## Verified Contract / Proposed Method / HOLD

**Verified Contract**
- target native head 使用 target train labels，因此名称必须是 transfer/adaptation；不得写成 zero-shot、zero-target 或 source-only generalization。
- source/target split、grouping 和 outer label isolation 必须逐 target 数据集独立冻结。

**Proposed Method**
- 每个 LODO fold：target 不参与 source representation/native-head training 或 selection；冻结 source-trained AST representation。
- 只用 target train/internal-validation labels 拟合 target native head；target outer/test 仅 terminal evaluation。
- 与 matched target-native/reference 比较，报告 adaptation budget 与 target-label count。

**HOLD**
- 四个 LODO target 的合法 source surface、source training recipe、target-head budget/seed/selection、target prerequisite parity 尚未冻结。
- target encoder fine-tuning、threshold search、shared KAUH mapping/diagnosis HOLD。
